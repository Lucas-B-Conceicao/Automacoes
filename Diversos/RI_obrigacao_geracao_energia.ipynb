```python
import pandas
import os
from datetime import datetime
from xlsxwriter import Workbook
```

```python
Caminho_arquivo_origem ='C:/Users/lucas.conceicao/OneDrive 2026/AXS ENERGIA S A/Gestão de Usinas - Usinas AXS/03.Análises/14 - Controle de Faturas e Recebimentos/Faturas/Medição - Unidade Geradora.xlsx'
Caminho_arquivo_destino = 'C:/Users/lucas.conceicao/Downloads'

dataframe = None
dataframe_tratado = None
dataframe_final = None
```

```python
def leitura ():

    global dataframe

    dataframe = pandas.read_excel(Caminho_arquivo_origem, skiprows=0, sheet_name="Suporte Medições")

    return dataframe
```

```python
def tratamento ():

    global dataframe_tratado

    dataframe_tratado = dataframe.drop(columns=["Compensação", "Consumo Fora Ponta", "Consumo Ponta",
                                                "Energia Injetada Ponta", "Energia Injetada Fora Ponta", "Saldo Anterior P kWh", "Saldo Anterior FP kWh"])
    return dataframe_tratado
```

```python
def filtros (ativo, distribuidora=None, periodo_inicial=None, periodo_final=None):

    global dataframe_final
    
    if periodo_inicial == None and periodo_final == None and distribuidora == None:

        dataframe_final = dataframe_tratado[(dataframe_tratado['CRI']==ativo)]

    elif periodo_inicial == None and periodo_final == None and distribuidora != None:
        
        dataframe_final = dataframe_tratado[(dataframe_tratado['CRI']==ativo) & (dataframe_tratado['Distribuidora']==distribuidora)]

    elif periodo_inicial != None and periodo_final != None and distribuidora == None:

        dataframe_final = dataframe_tratado[(dataframe_tratado['CRI']==ativo) & 
                                            (dataframe_tratado['Referência']>=periodo_inicial) & (dataframe_tratado['Referência']<=periodo_final)]
    elif ativo == None:

        dataframe_final = dataframe_tratado[(dataframe_tratado['Distribuidora']==distribuidora) & 
                                            (dataframe_tratado['Referência']>=periodo_inicial) & (dataframe_tratado['Referência']<=periodo_final)]
    else:
        dataframe_final = dataframe_tratado[(dataframe_tratado['CRI']==ativo) & (dataframe_tratado['Distribuidora']==distribuidora) & 
                                            (dataframe_tratado['Referência']>=periodo_inicial) & (dataframe_tratado['Referência']<=periodo_final)]
                                            
    print(dataframe_final)
    return dataframe_final
```

```python
def salvar (ativo, distribuidora=None, periodo_inicial=None, periodo_final=None):

    if periodo_inicial and periodo_final:
        inicial = periodo_inicial.replace("-","")[2:6]
        final = periodo_final.replace("-","")[2:6]

    if periodo_inicial == None and periodo_final == None and distribuidora == None:
        
        destino = "{}/{}.xlsx".format(Caminho_arquivo_destino, ativo)
    
    if periodo_inicial == None and periodo_final == None and distribuidora != None:

        destino = "{}/{} {}.xlsx".format(Caminho_arquivo_destino, ativo, distribuidora)

    if periodo_inicial != None and periodo_final != None and distribuidora == None:

        destino = "{}/{} {}-{}.xlsx".format(Caminho_arquivo_destino, ativo, inicial, final)

    if periodo_inicial != None and periodo_final != None and distribuidora != None:

        destino = "{}/{} {} {}-{}.xlsx".format(Caminho_arquivo_destino, ativo, distribuidora, inicial, final)

    if not os.path.exists(destino):
        with pandas.ExcelWriter(destino, engine='xlsxwriter') as writer:
            dataframe_final.to_excel(writer, sheet_name='Volume de Geração Total', index=False)

            planilha = writer.book

            aba_geral = writer.sheets['Volume de Geração Total']
            
            formato_ultima_coluna = planilha.add_format({'num_format': '#,##0.00'})

            aba_geral.set_column('F:F', None, formato_ultima_coluna)
```

```python
def principal (ativo, distribuidora, periodo_inicial, periodo_final):

    leitura()
    tratamento()
    filtros(ativo, distribuidora, periodo_inicial, periodo_final)
    salvar(ativo, distribuidora, periodo_inicial, periodo_final)
```
